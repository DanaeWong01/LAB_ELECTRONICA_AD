Reviso el archivo generado en scilab y genero una lista de valores de amplitud

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = open("seno.txt")
sen = f.readlines()
f.close()
x = np.linspace(1, len(sen),len(sen))
# Quito los sufijos \n
sen_def = []
for val in sen:
    sen_def.append(int(val[:-1]))
print(sen_def[0]+sen_def[1])
```

```python
plt.figure(figsize=(18,18))
plt.plot(x,sen_def)
plt.grid()
plt.show()
```

```python
with open("ROM.txt", "w") as arch:
    for num in range (len(sen_def)):
        arch.write(f"8'h{format(num, '02x')}:n_out = 9'b{sen_def[num]:08b};\n") 
```

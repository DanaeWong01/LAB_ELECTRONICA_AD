```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = open("seno.txt")
sen = f.readlines()
f.close()
x = np.linspace(1, len(sen),len(sen))
# Quito los sufijos \n
sen_def = []
for val in sen:
    sen_def.append(int(val[:-1]))
print(sen_def[0]+sen_def[1])
```

```python
plt.figure(figsize=(10,8))
plt.plot(x,sen_def)
plt.grid()
plt.show()
```

```python
with open("cuarto_sen.coe", "w") as f:

    f.write("memory_initialization_radix=16;\n")
    f.write("memory_initialization_vector=\n")

    for i, sample in enumerate(sen_def):

        if i == len(sen_def )-1:
            f.write(f"{sample:02X};\n")
        else:
            f.write(f"{sample:02X},\n")

print(sen_def)
```
